```python
# Standard Side-Chain van der Waals Volumes (in cubic Angstroms)
# Values approximate total residue volume minus standard backbone volume (~43.5)
VDW_VOLUME_DICT = {
    'G': 0.0,    # Glycine (No heavy atom side-chain)
    'A': 15.3,   # Alanine
    'S': 16.9,   # Serine
    'C': 42.6,   # Cysteine
    'D': 47.7,   # Aspartic Acid
    'P': 49.3,   # Proline
    'N': 50.8,   # Asparagine
    'T': 52.8,   # Threonine
    'E': 64.9,   # Glutamic Acid
    'V': 62.2,   # Valine
    'Q': 68.3,   # Glutamine
    'H': 80.0,   # Histidine
    'M': 81.3,   # Methionine
    'I': 85.0,   # Isoleucine
    'L': 85.0,   # Leucine
    'K': 97.4,   # Lysine
    'R': 114.3,  # Arginine
    'F': 113.8,  # Phenylalanine
    'Y': 130.3,  # Tyrosine
    'W': 149.7   # Tryptophan
}

VDW_VOLUME_DICT_2 = {
    'G': 0.0,    # Glycine (No heavy atom side-chain)
    'A': 25.6,   # Alanine
    'S': 30.6,   # Serine
    'C': 39.6,   # Cysteine
    'D': 52.1,   # Aspartic Acid
    'P': 58.7,   # Proline
    'N': 61.6,   # Asparagine
    'T': 56.7,   # Threonine
    'E': 74.9,   # Glutamic Acid
    'V': 74.3,   # Valine
    'Q': 85.3,   # Glutamine
    'H': 93.7,   # Histidine
    'M': 101.9,   # Methionine
    'I': 100.2,   # Isoleucine
    'L': 99.6,   # Leucine
    'K': 105.1,   # Lysine
    'R': 128.1,  # Arginine
    'F': 127.5,  # Phenylalanine
    'Y': 131.3,  # Tyrosine
    'W': 162.6   # Tryptophan
}
def calculate_cumulative_volume(sequence):
    """Calculates the total VdW side-chain volume for a given sub-register."""
    return sum(VDW_VOLUME_DICT.get(res, "A") for res in sequence)

def calculate_cumulative_volume2(sequence):
    """Calculates the total VdW side-chain volume for a given sub-register."""
    return sum(VDW_VOLUME_DICT_2.get(res, "A") for res in sequence)


af_pose = pyrosetta.pose_from_pdb("../pdb_files/AF-P00974-F1-model_v6.pdb")
af_pose.sequence()

def get_residue_sequence(residues_str):
    if pd.isna(residues_str) or residues_str == '':
        return ''

    residue_start_idx = int(residues_str.split('_')[0])  # Get the starting residue ID
    residue_end_idx = int(residues_str.split('_')[1])  # Get the ending residue ID

    residue_ids = range(residue_start_idx, residue_end_idx + 1)  # Create a range of residue IDs
    return ''.join(af_pose.residue(rid).name1() for rid in residue_ids)

def calculate_cumulative_chi_2(residues):
    start_idx = int(residues.split('_')[0])
    end_idx = int(residues.split('_')[1])

    cumulative_angles = 0
    for res_id in range(start_idx, end_idx + 1):
        chi_anlges = af_pose.residue(res_id).nchi()
        cumulative_angles += chi_anlges
    return cumulative_angles

from math import pi
def calculate_cumulative_volume3(residues):
    """Calculates the total VdW side-chain volume for a given sub-register."""
    start_idx = int(residues.split('_')[0])
    end_idx = int(residues.split('_')[1])

    cumulative_volume = 0
    for res_id in range(start_idx, end_idx + 1):
        residue_radius = af_pose.residue(res_id).nbr_radius()
        cur_vol = (4.0 / 3.0) * pi * (residue_radius ** 3)
        cumulative_volume += cur_vol
    return cumulative_volume


def calculate_cumulative_volume4(residues):
    start_idx = int(residues.split('_')[0])
    end_idx = int(residues.split('_')[1])

    cumulative_volume = 0
    for res_id in range(start_idx, end_idx + 1):
        residue = af_pose.residue(res_id)
        res_volume = 0.0

        type_set = residue.atom_type_set()

        for j in range(1, residue.natoms() + 1):
            # Retrieve the atom type index and its associated volume
            atom_type_index = residue.atom_type_index(j)
            atom_type = type_set[atom_type_index]

            # Use lj_radius to calculate volume of a sphere: (4/3) * pi * r^3
            radius = atom_type.lj_radius()
            atom_volume = (4.0 / 3.0) * pi * (radius ** 3)

            res_volume += atom_volume


            # .vol() returns the VDW volume defined in the Rosetta database
        cumulative_volume += res_volume
    return cumulative_volume

```

```python
for df in [low_qubit_confident_df, high_qubit_confident_df, low_qubit_unconfident_df]:
    df['repacked_sequence'] = df['residues'].apply(get_residue_sequence)
    df['cumulative_vdw_volume_1'] = df['repacked_sequence'].apply(calculate_cumulative_volume)
    df['cumulative_vdw_volume_2'] = df['repacked_sequence'].apply(calculate_cumulative_volume2)
    df['avg_vdw_volume_1'] = df['cumulative_vdw_volume_1'] / df['residue_count']
    df['avg_vdw_volume_2'] = df['cumulative_vdw_volume_2'] / df['residue_count']
    df['chi_dof'] = df['residues'].apply(calculate_cumulative_chi_2)
    df['avg_chi_dof'] = df['chi_dof'] / df['residue_count']
```

```python

 # Confidence of low qubit - high pLDDT confidence results
def compute_correlation_and_confidence(df, feature):
    r, p_pearson = stats.pearsonr(df[feature], df['energy_diff'])
    rho, p_spearman = stats.spearmanr(df[feature], df['energy_diff'])

    return r, p_pearson, rho, p_spearman, abs(r - rho)

def compute_all_correlations(df, features=None):
    if features is None:
        features = ['residue_count', 'num_qubits', 'avg_chi_dof', 'avg_vdw_volume_2']
    results = {}

    for feature in features:
        r, p_pearson, rho, p_spearman, confidence_delta = compute_correlation_and_confidence(df, feature)
        results[feature] = {
            'pearson_r': r,
            'pearson_p': p_pearson,
            'spearman_rho': rho,
            'spearman_p': p_spearman,
            'confidence_delta': confidence_delta
        }
    return results

def compute_correlations_for_all_dfs(dfs, features=None):
    all_results = {}
    for name, df in dfs.items():
        all_results[name] = compute_all_correlations(df, features)
    return all_results
```
