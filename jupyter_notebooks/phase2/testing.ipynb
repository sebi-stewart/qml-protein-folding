```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

import pathlib

sns.set_style("whitegrid")
sns.set_context("paper")
sns.set_palette("deep", color_codes=True)
```

```python
files = [
    "phase2_results.pkl",
    "phase2_11_to_14_results.pkl",
    "phase2_18_qubits_high_confidence_results.pkl",
    "phase2_22_qubits_high_confidence_results.pkl",
]
combined_df = pd.concat(
    [pd.read_pickle(file) for file in files],
    ignore_index=True,
)
combined_df
```

```python
one_hot_classifications = pd.get_dummies(combined_df.classification)
combined_df = pd.concat([combined_df, one_hot_classifications], axis=1)

combined_df["starting_residue"] = combined_df.residues.apply(lambda x: int(x.split("_")[0]))
combined_df["ending_residue"] = combined_df.residues.apply(lambda x: int(x.split("_")[1]))


grouped_df = combined_df.groupby(["residues", "residue_count"]).agg({"energy_diff": ["min", "max", 'mean', "std"]}).reset_index()

grouped_df["starting_residue"] = grouped_df.residues.apply(lambda x: int(x.split("_")[0]))
grouped_df["ending_residue"] = grouped_df.residues.apply(lambda x: int(x.split("_")[1]))

grouped_df.columns = [
    "_".join(col).strip("_") if isinstance(col, tuple) else col
    for col in grouped_df.columns
]
grouped_df
```

```python
combined_df.energy_diff.nunique()
```

```python
reside_counts = grouped_df.residue_count.unique()
fig, axes = plt.subplots(len(reside_counts), 1, figsize=(10, 6 * len(reside_counts)), sharex=True)

for i, count in enumerate(sorted(reside_counts)):
    subset = grouped_df[grouped_df.residue_count == count]
    axes[i].errorbar(
        subset.starting_residue,
        subset.energy_diff["mean"],
        yerr=subset.energy_diff["std"],
        fmt="o",
        ecolor="gray",
        capsize=5,
        label=f"{count} residues",
    )
    axes[i].set_title(f"Energy Difference for {count} Residues")
    axes[i].set_ylabel("Energy Difference (QAOA - Classical)")
    axes[i].legend()
    axes[i].grid(True)
```

```python
# Plot the differing residue lengths on the same plot with different colors
plt.figure(figsize=(12, 8))
for count in sorted(reside_counts):
    subset = grouped_df[grouped_df.residue_count == count]
    plt.errorbar(
        subset.starting_residue,
        subset.energy_diff["mean"],
        yerr=subset.energy_diff["std"],
        fmt="o",
        ecolor="gray",
        capsize=5,
        label=f"{count} residues",
    )
plt.title("Energy Difference by Starting Residue for Different Residue Counts")
plt.xlabel("Starting Residue")
plt.ylabel("Energy Difference (QAOA - Classical)")
plt.legend()
plt.grid(True)
plt.tight_layout()
# plt.savefig("../plots/energy_diff_by_residue_count.svg", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Add column to combined_df for qubit count category
def qubit_category(num_qubits):
    if num_qubits <= 14:
        return "<=14"
    elif num_qubits == 18:
        return "18"
    elif num_qubits == 22:
        return "22"
combined_df["qubit_category"] = combined_df.num_qubits.apply(qubit_category)
```

```python
from matplotlib.ticker import FuncFormatter

order = ["<=14", "18", "22"]
hue_order = ["loss", "tie", "win"]
palette = {"win": "#2ca02c", "tie": "#1f77b4", "loss": "#d62728"}

fig, ax = plt.subplots(figsize=(8, 5))

sns.histplot(
    data=combined_df,
    x="qubit_category",
    hue="classification",

    # order=order,
    hue_order=["loss", "tie", "win"],
    palette=palette,

    discrete=True,
    shrink=0.8,
    multiple="fill",
    stat="count",
    ax=ax,
)

ax.yaxis.set_major_formatter(FuncFormatter(lambda y, _: f"{y*100:.0f}%"))
ax.set_ylabel("Percent within qubit category", fontsize=16)
ax.set_xlabel("Qubit category", fontsize=16)
ax.set_ylim(0, 1)
ax.set_title("Classification Percentages by Qubit Category", fontsize=18)
ax.legend(title="Classification", labels=["Wins", "Ties", "Losses"], fontsize=14, title_fontsize=15)
ax.tick_params(axis="both", labelsize=14)

for p in ax.patches:
    h = p.get_height()
    if np.isnan(h) or h <= 0:
        continue

    # Only label meaningful segments to avoid clutter (e.g., >5%)
    if h >= 0.05:
        x = p.get_x() + p.get_width() / 2
        y = p.get_y() + h / 2
        ax.text(
            x, y, f"{h*100:.1f}%",
            ha="center", va="center",
            color="white", fontsize=14, fontweight="bold"
        )

# Annotate the tiny 4.5% segment (<=14 wins) with an arrow from below
for p in ax.patches:
    h = p.get_height()
    if np.isnan(h) or h <= 0:
        continue
    if 0.03 < h < 0.05:
        x = p.get_x() + p.get_width() / 2
        y_tip = p.get_y() + h / 2
        ax.annotate(
            f"{h*100:.1f}%",
            xy=(x, y_tip),
            xytext=(x, y_tip - 0.09),
            ha="center", va="top",
            color="white", fontsize=13, fontweight="bold",
            bbox=dict(boxstyle="round,pad=0.3", fc="#d62728", ec="none"),
            arrowprops=dict(arrowstyle="-|>", color="#d62728", lw=2.5),
        )

plt.tight_layout()

plt.savefig("../../Publication/plots/classification_percentages_by_qubit_category.svg", dpi=300, bbox_inches="tight")
plt.show()
```
