## Parameters

| Parameter                   | Description                                                                       |
|-----------------------------|-----------------------------------------------------------------------------------|
| `SOURCE_FOLDER`             | Path (relative to project root) containing energy mapping `.npz` files            |
| `START_AT_FILE_IDX`         | Skip the first N files — useful when running multiple Colab instances in parallel |
| `LIMIT_FILES_PER_QUBIT`     | Max files to process per qubit count                                              |
| `QUBIT_COUNTS_TO_ANALYSE`   | List of qubit counts to include                                                   |
| `BASE_EPOCHS`               | Optimiser iterations per QAOA layer                                               |
| `BASE_STEPSIZE`             | Initial gradient-descent step size                                                |
| `QAOA_LAYERS`               | List of layer depths to sweep over                                                |
| `STEPSIZE_REDUCTION_FACTOR` | Step size is divided by this for each successive layer                            |
| `HIGH_TO_LOW_QUBIT_ORDER`   | Process largest qubit counts first if `True`                                      |
| `USE_GPU`                   | Use `lightning.gpu` backend if `True`, otherwise `lightning.qubit`                |
| `QAOA_SEED_COUNT`           | Number of random seeds per run                                                    |
| `QAOA_EXECUTION_MODE`       | `"batched"` (faster, more memory) or `"sequential"`                               |
| `QAOA_MAX_MEMORY_GB`        | Memory cap in GB for batched execution                                            |
| `OUTPUTS_FOLDER`            | Where result `.npz` files are written                                             |
| `LOGS_FOLDER`               | Where log files are written                                                       |
| `LOGS_FILE`                 | Log file base name                                                                |
